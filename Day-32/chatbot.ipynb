```python
import google.generativeai as genai
from google.colab import userdata

# Get your Gemini API key from Colab's secret manager
# Create a secret named 'GEMINI_API_KEY' in Colab's Secret Manager
# if you haven't already. You can get a free API key from https://aistudio.google.com/app/apikey
try:
    API_KEY = userdata.get('GEMINI_API_KEY')
    genai.configure(api_key=API_KEY)

    # Initialize the Generative Model with 'gemini-2.5-flash'
    model = genai.GenerativeModel('gemini-2.5-flash')

    print("Gemini Chatbot (type exit to quit)\n")
    messages = [] # Initialize messages list for conversational context

    while True:
        user_input = input("You: ")

        if user_input.lower() == "exit":
            print("Bot: Goodbye!")
            break

        messages.append({"role": "user", "parts": [user_input]})

        response = model.generate_content(messages)

        reply = response.text
        print("Bot:", reply)

        messages.append({"role": "model", "parts": [reply]})

except Exception as e:
    print(f"An error occurred: {e}")
    print("Please ensure you have set up your 'GEMINI_API_KEY' in Colab's Secret Manager.")
    print("You can get a free API key from https://aistudio.google.com/app/apikey")
```
